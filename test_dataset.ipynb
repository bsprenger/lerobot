```python
from copy import deepcopy

import torch

from lerobot.common.datasets.compute_stats import aggregate_stats, compute_episode_stats
from lerobot.common.datasets.lerobot_dataset import LeRobotDataset
from lerobot.common.datasets.utils import get_episode_data_index

# TODO: should it be inplace? or should it return a new dataset?
# what would be conventional based on other libs?
# TODO implement in such a way that if anything fails, all rolls back
# first: modify the hf_dataset


def remove_episodes(dataset: LeRobotDataset, episodes_to_remove: list[int]) -> None:  # TODO check return type
    # --- MODIFYING DATA --- #
    episodes_to_keep = [i for i in range(dataset.meta.total_episodes) if i not in episodes_to_remove]
    remaining_episode_indices = sorted(episodes_to_keep)
    old_to_new = {old: new for new, old in enumerate(remaining_episode_indices)}

    def filter_and_map_fn(batch, indices):
        # Create a mask for episodes to keep
        # 1 if keeping that row, 0 if removing
        keep_mask = [ep_idx not in episodes_to_remove for ep_idx in batch["episode_index"]]
        if not any(keep_mask):
            return {k: [] for k in batch}

        # Filter all columns based on the keep mask
        filtered_batch = {k: [v for v, keep in zip(batch[k], keep_mask, strict=False) if keep] for k in batch}

        # Update the episode index
        filtered_batch["episode_index"] = [
            old_to_new[ep_idx.item()] for ep_idx in filtered_batch["episode_index"]
        ]

        # may also need to update the 'index' key but not sure what that is about yet
        return filtered_batch

    new_hf_dataset = dataset.hf_dataset.map(
        filter_and_map_fn,
        batched=True,
        with_indices=True,
        remove_columns=dataset.hf_dataset.column_names,
    )
    # --- --- --- #

    # --- MODIFYING META --- #
    new_meta = deepcopy(dataset.meta)
    new_meta.info["total_episodes"] = len(remaining_episode_indices)
    new_meta.info["total_frames"] = len(new_hf_dataset)
    task_indices = set()
    for ep_idx in remaining_episode_indices:
        if ep_idx in dataset.meta.episodes:
            for task_idx in dataset.meta.episodes[ep_idx]["tasks"]:
                task_indices.add(task_idx)
    new_meta.info["total_tasks"] = len(task_indices)
    new_meta.info["total_videos"] = (
        len(remaining_episode_indices) * len(dataset.meta.video_keys) if dataset.meta.video_keys else 0
    )
    new_meta.info["total_chunks"] = None  # TODO

    new_meta.tasks = ...  # TODO

    new_episodes = {k: v for k, v in new_meta.episodes.items() if k not in episodes_to_remove}
    new_episodes = {i: {**v, "episode_index": i} for i, v in enumerate(new_episodes.values())}
    new_meta.episodes = new_episodes

    new_episode_data_index = get_episode_data_index(new_episodes)
    episode_stats = {}  # TODO what are the keys here supposed to be?
    for from_idx, to_idx, episode in zip(
        new_episode_data_index["from"], new_episode_data_index["to"], new_episodes.values(), strict=False
    ):
        episode_buffer = {
            k: torch.stack(new_hf_dataset[from_idx:to_idx][k]).numpy() for k in new_hf_dataset[0]
        }
        ep_stats = compute_episode_stats(episode_buffer, new_meta.features)
        episode_stats[episode["episode_index"]] = ep_stats  # TODO not sure about this
    new_meta.episodes_stats = episode_stats  # depends on version write_episode_stats for each episode in dict
    new_meta.stats = aggregate_stats(new_meta.episodes_stats)

    # MODIFYING VIDEOS
    vid_keys = dataset.meta.video_keys
    if vid_keys:
        for old, new in old_to_new.items():
            vid_keys = dataset.meta.video_keys
            old_paths = [dataset.meta.get_video_file_path(old, k) for k in vid_keys]
            for old_path in old_paths:
                # Since we should only overwrite with existing files, we can get
                # the new path this way
                new_path = dataset.meta.get_video_file_path(new, k)
                old_path.rename(new_path)

    # Now save everything to file

    return dataset


repo_id = "lerobot/aloha_mobile_cabinet"
dataset = LeRobotDataset(repo_id)

remove_episodes(dataset, [0, 1, 2])
```
